```python
# ============================================================
# TASK 1 — Unit Converter
# ============================================================

def convert_length(value, direction):
    """
    Converts between meters and feet.
    Parameters:
        value (float): The value to convert.
        direction (str): '1' = meters to feet, '2' = feet to meters.
    Returns:
        float: Converted value.
    """
    if direction == '1':
        result = value * 3.28084
        print(f"{value} meters = {result:.4f} feet")
    elif direction == '2':
        result = value / 3.28084
        print(f"{value} feet = {result:.4f} meters")
    else:
        print("Invalid direction.")
        return None
    return result


def convert_weight(value, direction):
    """
    Converts between kilograms and pounds.
    Parameters:
        value (float): The value to convert.
        direction (str): '1' = kg to lbs, '2' = lbs to kg.
    Returns:
        float: Converted value.
    """
    if direction == '1':
        result = value * 2.20462
        print(f"{value} kg = {result:.4f} lbs")
    elif direction == '2':
        result = value / 2.20462
        print(f"{value} lbs = {result:.4f} kg")
    else:
        print("Invalid direction.")
        return None
    return result


def convert_volume(value, direction):
    """
    Converts between liters and gallons.
    Parameters:
        value (float): The value to convert.
        direction (str): '1' = liters to gallons, '2' = gallons to liters.
    Returns:
        float: Converted value.
    """
    if direction == '1':
        result = value * 0.264172
        print(f"{value} liters = {result:.4f} gallons")
    elif direction == '2':
        result = value / 0.264172
        print(f"{value} gallons = {result:.4f} liters")
    else:
        print("Invalid direction.")
        return None
    return result


def unit_converter():
    """
    Main unit converter. Prompts user to pick a category and direction,
    then performs the conversion with full error handling.
    Returns:
        float: The converted value, or None on invalid input.
    """
    print("========== Unit Converter ==========")
    print("1. Length  (meters <-> feet)")
    print("2. Weight  (kg <-> lbs)")
    print("3. Volume  (liters <-> gallons)")
    category = input("Enter 1, 2, or 3: ").strip()

    if category not in ['1', '2', '3']:
        print("Invalid category.")
        return None

    if category == '1':
        print("1 = meters to feet | 2 = feet to meters")
    elif category == '2':
        print("1 = kg to lbs      | 2 = lbs to kg")
    elif category == '3':
        print("1 = liters to gal  | 2 = gal to liters")

    direction = input("Enter 1 or 2: ").strip()

    try:
        value = float(input("Enter value to convert: "))
    except ValueError:
        print("Invalid input! Please enter a number.")
        return None

    if category == '1':
        return convert_length(value, direction)
    elif category == '2':
        return convert_weight(value, direction)
    elif category == '3':
        return convert_volume(value, direction)

unit_converter()
```

```python
# ============================================================
# TASK 2 — Math Operations on a List
# ============================================================

def calculate_sum(numbers):
    """
    Returns the sum of a list of numbers.
    Parameters:
        numbers (list): List of numeric values.
    Returns:
        float: Sum of all elements.
    """
    total = 0
    for n in numbers:
        total += n
    return total


def calculate_average(numbers):
    """
    Returns the average of a list of numbers.
    Parameters:
        numbers (list): List of numeric values.
    Returns:
        float: Average of all elements.
    """
    return calculate_sum(numbers) / len(numbers)


def calculate_maximum(numbers):
    """
    Returns the maximum value in a list.
    Parameters:
        numbers (list): List of numeric values.
    Returns:
        float: Largest element.
    """
    max_val = numbers[0]
    for n in numbers:
        if n > max_val:
            max_val = n
    return max_val


def calculate_minimum(numbers):
    """
    Returns the minimum value in a list.
    Parameters:
        numbers (list): List of numeric values.
    Returns:
        float: Smallest element.
    """
    min_val = numbers[0]
    for n in numbers:
        if n < min_val:
            min_val = n
    return min_val


def math_operations():
    """
    Prompts user to pick an operation and input numbers,
    then performs and displays the result with error handling.
    Returns:
        float: Result of the selected operation, or None on error.
    """
    print("========== Math Operations ==========")
    print("1. Sum")
    print("2. Average")
    print("3. Maximum")
    print("4. Minimum")
    operation = input("Enter 1, 2, 3, or 4: ").strip()

    if operation not in ['1', '2', '3', '4']:
        print("Invalid operation.")
        return None

    raw = input("Enter numbers separated by spaces (e.g. 1 2 3 4): ")

    try:
        numbers = [float(x) for x in raw.strip().split()]
        if len(numbers) == 0:
            print("Error: Empty list!")
            return None
    except ValueError:
        print("Invalid input! Numbers only please.")
        return None

    if operation == '1':
        result = calculate_sum(numbers)
        print(f"Sum: {result}")
    elif operation == '2':
        result = calculate_average(numbers)
        print(f"Average: {result:.4f}")
    elif operation == '3':
        result = calculate_maximum(numbers)
        print(f"Maximum: {result}")
    elif operation == '4':
        result = calculate_minimum(numbers)
        print(f"Minimum: {result}")

    return result

math_operations()
```

```python
# ============================================================
# SECTION 4.2 — List Manipulation (Tasks 1-7)
# ============================================================

def extract_every_other(lst):
    """
    Extracts every other element starting from index 0.
    Parameters:
        lst (list): Input list.
    Returns:
        list: Elements at indices 0, 2, 4...
    """
    return lst[::2]

def get_sublist(lst, start, end):
    """
    Returns sublist from start to end index (inclusive).
    Parameters:
        lst (list): Input list.
        start (int): Start index.
        end (int): End index (inclusive).
    Returns:
        list: Sublist from start to end.
    """
    return lst[start:end + 1]

def reverse_list(lst):
    """
    Reverses a list using slicing.
    Parameters:
        lst (list): Input list.
    Returns:
        list: Reversed list.
    """
    return lst[::-1]

def remove_first_last(lst):
    """
    Removes first and last elements using slicing.
    Parameters:
        lst (list): Input list.
    Returns:
        list: List without first and last elements.
    """
    return lst[1:-1]

def get_first_n(lst, n):
    """
    Returns the first n elements of a list.
    Parameters:
        lst (list): Input list.
        n (int): Number of elements from the start.
    Returns:
        list: First n elements.
    """
    return lst[:n]

def get_last_n(lst, n):
    """
    Returns the last n elements of a list.
    Parameters:
        lst (list): Input list.
        n (int): Number of elements from the end.
    Returns:
        list: Last n elements.
    """
    return lst[-n:]

def reverse_skip(lst):
    """
    Returns every second element starting from second-to-last, going backwards.
    Parameters:
        lst (list): Input list.
    Returns:
        list: Every second element in reverse order from index -2.
    """
    return lst[-2::-2]


# ---- Tests ----
print("Task 1 - Every other:      ", extract_every_other([1, 2, 3, 4, 5, 6]))   # [1, 3, 5]
print("Task 2 - Sublist:          ", get_sublist([1, 2, 3, 4, 5, 6], 2, 4))     # [3, 4, 5]
print("Task 3 - Reverse:          ", reverse_list([1, 2, 3, 4, 5]))              # [5, 4, 3, 2, 1]
print("Task 4 - Remove first/last:", remove_first_last([1, 2, 3, 4, 5]))         # [2, 3, 4]
print("Task 5 - First 3:          ", get_first_n([1, 2, 3, 4, 5], 3))            # [1, 2, 3]
print("Task 6 - Last 2:           ", get_last_n([1, 2, 3, 4, 5], 2))             # [4, 5]
print("Task 7 - Reverse skip:     ", reverse_skip([1, 2, 3, 4, 5, 6]))           # [5, 3, 1]
```

```python
# ============================================================
# SECTION 4.3 — Nested List Operations (Tasks 1-8)
# ============================================================

def flatten(lst):
    """
    Flattens one level of nesting into a single list.
    Parameters:
        lst (list): Nested list (one level deep).
    Returns:
        list: Flat list of all elements.
    """
    flat = []
    for sublist in lst:
        for item in sublist:
            flat.append(item)
    return flat

def access_nested_element(lst, indices):
    """
    Accesses an element in a nested list using a list of indices.
    Parameters:
        lst (list): Nested list.
        indices (list): List of indices to navigate into the nested list.
    Returns:
        The element at the given nested position.
    """
    element = lst
    for i in indices:
        element = element[i]
    return element

def sum_nested(lst):
    """
    Sums all numbers in a nested list regardless of depth using recursion.
    Parameters:
        lst (list): Nested list of numbers.
    Returns:
        float: Total sum of all elements.
    """
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)
        else:
            total += item
    return total

def remove_element(lst, elem):
    """
    Removes all occurrences of an element from a nested list.
    Parameters:
        lst (list): Nested list.
        elem: Element to remove.
    Returns:
        list: Nested list with all occurrences of elem removed.
    """
    return [[item for item in sublist if item != elem] for sublist in lst]

def find_max(lst):
    """
    Finds the maximum element in a nested list regardless of depth.
    Parameters:
        lst (list): Nested list of numbers.
    Returns:
        float: Maximum value found.
    """
    max_val = None
    for item in lst:
        candidate = find_max(item) if isinstance(item, list) else item
        if max_val is None or candidate > max_val:
            max_val = candidate
    return max_val

def count_occurrences(lst, elem):
    """
    Counts how many times an element appears in a nested list.
    Parameters:
        lst (list): Nested list.
        elem: Element to count.
    Returns:
        int: Total count of elem.
    """
    count = 0
    for sublist in lst:
        for item in sublist:
            if item == elem:
                count += 1
    return count

def deep_flatten(lst):
    """
    Deeply flattens a list of any nesting depth using recursion.
    Parameters:
        lst (list): Deeply nested list.
    Returns:
        list: Single flat list of all elements.
    """
    flat = []
    for item in lst:
        if isinstance(item, list):
            flat.extend(deep_flatten(item))
        else:
            flat.append(item)
    return flat

def average_nested(lst):
    """
    Calculates the average of all elements in a nested list.
    Parameters:
        lst (list): Nested list of numbers.
    Returns:
        float: Average of all elements.
    """
    flat = deep_flatten(lst)
    return sum(flat) / len(flat)


# ---- Tests ----
print("Nested Task 1 - Flatten:          ", flatten([[1, 2], [3, 4], [5]]))
print("Nested Task 2 - Access element:   ", access_nested_element([[1,2,3],[4,5,6],[7,8,9]], [1, 2]))
print("Nested Task 3 - Sum nested:       ", sum_nested([[1, 2], [3, [4, 5]], 6]))
print("Nested Task 4 - Remove element:   ", remove_element([[1, 2], [3, 2], [4, 5]], 2))
print("Nested Task 5 - Find max:         ", find_max([[1, 2], [3, [4, 5]], 6]))
print("Nested Task 6 - Count occurrences:", count_occurrences([[1, 2], [2, 3], [2, 4]], 2))
print("Nested Task 7 - Deep flatten:     ", deep_flatten([[[1, 2], [3, 4]], [[5, 6], [7, 8]]]))
print("Nested Task 8 - Average nested:   ", average_nested([[1, 2], [3, 4], [5, 6]]))
```
